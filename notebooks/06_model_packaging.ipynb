```python
# ============================================================
# STEP 257 — IDENTIFY EXACT DDOS TABLE
# ============================================================

import json
from pathlib import Path

schema_path = (
    project_root
    / "models"
    / "dos_feature_schema.json"
)

with open(
    schema_path,
    "r",
    encoding="utf-8"
) as f:
    schema = json.load(f)

required_dos_features = schema["features"]

print(
    "Required DDoS features:",
    len(required_dos_features)
)

for name, obj in list(globals().items()):

    if isinstance(obj, pd.DataFrame):

        feature_matches = [
            col
            for col in required_dos_features
            if col in obj.columns
        ]

        if len(feature_matches) >= 60:

            print("\n✅ Candidate:", name)

            print(
                "Shape:",
                obj.shape
            )

            print(
                "DDoS features present:",
                len(feature_matches),
                "/",
                len(required_dos_features)
            )

            print(
                "Has SrcAddr:",
                "SrcAddr" in obj.columns
            )

            print(
                "Has time_window:",
                "time_window" in obj.columns
            )
```
